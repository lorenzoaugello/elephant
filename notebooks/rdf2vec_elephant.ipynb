```python
%%writefile /content/run_rdf2vec.py

from rdflib import Graph, Namespace, RDF
from pyrdf2vec import RDF2VecTransformer
from pyrdf2vec.embedders import Word2Vec
from pyrdf2vec.graphs import KG
from pyrdf2vec.walkers import RandomWalker


wordnet_file = "/content/english-wordnet-2025.ttl"
animal_file = "/content/animals_oewn_rdf (6).ttl"

WN = Namespace("https://globalwordnet.github.io/schemas/wn#")
ONTOLEX = Namespace("http://www.w3.org/ns/lemon/ontolex#")


# 1. Build a WordNet taxonomy graph
full_graph = Graph()
full_graph.parse(wordnet_file, format="turtle")

structural_graph = Graph()

for s, o in full_graph.subject_objects(WN.hypernym):
    structural_graph.add((s, WN.hypernym, o))

for s, o in full_graph.subject_objects(WN.hyponym):
    structural_graph.add((s, WN.hyponym, o))

structural_graph.serialize(
    "/content/wordnet_taxonomy.ttl",
    format="turtle"
)


# 2. Get the selected animal synsets
animals = Graph()
animals.parse(animal_file, format="turtle")

entities = [
    str(s)
    for s in animals.subjects(RDF.type, ONTOLEX.LexicalConcept)
]

print("Animal synsets:", len(entities))


# 3. Use the full WordNet taxonomy as RDF2Vec context
kg = KG("/content/wordnet_taxonomy.ttl")


# 4. RDF2Vec
transformer = RDF2VecTransformer(
    Word2Vec(epochs=10),
    walkers=[
        RandomWalker(
            4,
            10,
            with_reverse=True
        )
    ],
)


# 5. Generate embeddings for the selected animal synsets
embeddings, _ = transformer.fit_transform(
    kg,
    entities
)

print("Animal synsets:", len(entities))
print("Number of embeddings:", len(embeddings))
print("Vector length:", len(embeddings[0]))

#save embeddings
import pickle

with open("/content/rdf2vec_embeddings.pkl", "wb") as f:
    pickle.dump(
        {
            "entities": entities,
            "embeddings": embeddings
        },
        f
    )

print("Embeddings saved to /content/rdf2vec_embeddings.pkl")

import csv

embedding_output = "/content/rdf2vec_embeddings_depth4.csv"

with open(
    embedding_output,
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.writer(f)

    # Header
    writer.writerow(
        ["synset_uri"] +
        [f"dim_{i}" for i in range(len(embeddings[0]))]
    )

    # One row per synset
    for entity, vector in zip(entities, embeddings):
        writer.writerow([entity] + list(vector))

print("Embeddings saved to:", embedding_output)
```

```python
%%writefile /content/compute_pair_similarities.py

import csv
import pickle
import numpy as np

from rdflib import Graph, Namespace


animal_file = "/content/animals_oewn_rdf (5).ttl"
embedding_file = "/content/rdf2vec_embeddings.pkl"

OEWN = Namespace("https://en-word.net/ontology/")
ONTOLEX = Namespace("http://www.w3.org/ns/lemon/ontolex#")


# 1. Load embeddings
with open(embedding_file, "rb") as f:
    data = pickle.load(f)

entities = data["entities"]
embeddings = data["embeddings"]

vectors = {
    entity: np.array(vector)
    for entity, vector in zip(entities, embeddings)
}


# 2. Load animal RDF
animals = Graph()
animals.parse(animal_file, format="turtle")


# 3. Map synset URI -> lexical label
labels = {}

for entry, synset in animals.subject_objects(ONTOLEX.evokes):
    for label in animals.objects(entry, ONTOLEX.writtenRep):
        labels.setdefault(str(synset), str(label))


# 4. Cosine similarity
def cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (
        np.linalg.norm(v1) * np.linalg.norm(v2)
    )


results = []


# 5. Target-hypernym pairs
for target, hypernym in animals.subject_objects(OEWN.hypernym):

    target = str(target)
    hypernym = str(hypernym)

    if target in vectors and hypernym in vectors:

        similarity = cosine_similarity(vectors[target], vectors[hypernym])

        results.append([
            f"{labels.get(target, target)} - {labels.get(hypernym, hypernym)}",
            "hypernym",
            f"{target} - {hypernym}",
            similarity
        ])


# 6. Target-hyponym pairs
for target, hyponym in animals.subject_objects(OEWN.hyponym):

    target = str(target)
    hyponym = str(hyponym)

    if target in vectors and hyponym in vectors:

        similarity = cosine_similarity(
            vectors[target],
            vectors[hyponym]
        )

        results.append([
            f"{labels.get(target, target)} - {labels.get(hyponym, hyponym)}",
            "hyponym",
            f"{target} - {hyponym}",
            similarity
        ])


# 7. Save CSV
output_file = "/content/rdf2vec_pair_similarities_depth4.csv"

with open(
    output_file,
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.writer(f)

    writer.writerow([
        "couple of words",
        "relation",
        "synset pair",
        "cosine similarity"
    ])

    writer.writerows(results)


print("Pairs written:", len(results))
print("Saved to:", output_file)
```
